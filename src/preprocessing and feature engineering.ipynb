# PART 4 - Feature Engineering

***Step 29 - Convert Appointment Date to Date Format***

```python
df["appointment_date"] = pd.to_datetime(df["appointment_date"])
```

***Step 30 - Extract Appointment Month And Appointment Date***

```python
df["appointment_month"] = df["appointment_date"].dt.month
df["appointment_day"] = df["appointment_date"].dt.day_name()
```

***Step 31 - Total Previous Visits***

```python
df["total_previous_visits"] = (
    df["previous_appointments"] +
    df["previous_admissions"]
)
```

***Step 32 - Previous No-Show Rate***

```python
df["previous_no_show_rate"] = (
    df["missed_previous_appointments"] /
    (df["previous_appointments"] + 1)
)
```

***Step 33 - Blood Pressure Difference***

```python
df["bp_difference"] = (
    df["systolic_bp"] -
    df["diastolic_bp"]
)
```

***Step 34 - Total Medical Charges***

```python
df["medical_cost"] = (
    df["consultation_fee_lkr"] +
    df["lab_charge_lkr"] +
    df["medicine_charge_lkr"]
)
```

***Step 35 - Average Cost per Treatment***

```python
df["average_treatment_cost"] = (
    df["total_bill_lkr"] /
    (df["treatments_count"] + 1)
)
```

***Step 36 - Remove Original Date Column***

```python
df.drop(columns=["appointment_date"], inplace=True)
```

# PART 5 - Feature Encoding and Scaling

***Step 38 - Display Categorical Features***

```python
categorical_columns = df.select_dtypes(include="object").columns

print("Categorical features:\n")

for col in categorical_columns:
    print(col)
```

***Step 39 - Label Encoding***

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

binary_columns = [
    "gender",
    "appointment_status",
    "admitted",
    "payment_status",
    "no_show",
    "readmitted_30_days"
]

for column in binary_columns:
    df[column] = label_encoder.fit_transform(df[column])
```

***Step 40 - One-Hot Encoding***

```python
df = pd.get_dummies(
    df,
    columns=[
        "blood_group",
        "department",
        "diagnosis",
        "payment_method",
        "appointment_day"
    ],
    drop_first=True
)
```

***Step 41 - Verify Target Variable***

```python
# Display original classes

print("Original Classes:")
print(df["disease_risk_level"].unique())

# Encode target variable

risk_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

df["disease_risk_level"] = df["disease_risk_level"].replace(risk_mapping)

# Verify encoding

print("Encoded Classes:")
print(sorted(df["disease_risk_level"].unique()))

print("\nClass Distribution:")

display(
    df["disease_risk_level"]
      .value_counts()
      .sort_index()
      .rename_axis("Disease Risk Level")
      .reset_index(name="Count")
)
```

***Step 42 - Feature Scaling***

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

columns_to_scale = [
    "age",
    "waiting_days",
    "previous_appointments",
    "missed_previous_appointments",
    "length_of_stay_days",
    "previous_admissions",
    "systolic_bp",
    "diastolic_bp",
    "blood_sugar_mg_dl",
    "cholesterol_mg_dl",
    "bmi",
    "lab_tests_count",
    "treatments_count",
    "consultation_fee_lkr",
    "room_charge_lkr",
    "lab_charge_lkr",
    "medicine_charge_lkr",
    "total_bill_lkr",
    "appointment_month",
    "total_previous_visits",
    "previous_no_show_rate",
    "bp_difference",
    "medical_cost",
    "average_treatment_cost"
]

df[columns_to_scale] = scaler.fit_transform(df[columns_to_scale])
```
